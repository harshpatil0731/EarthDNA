# EarthDNA 2024 Daily AOI-Level Feature EDA

This notebook examines the Milestone 1.4 validation/prototype feature tables. It does not train a model or define the final multi-year study dataset.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'experiments' else Path.cwd()
REGIONS = ('uttarakhand', 'california', 'australia')
FEATURES = {}
for region in REGIONS:
    path = ROOT / 'datasets' / 'processed' / region / 'daily_aoi_features_2024-01-01_2024-12-31.csv'
    frame = pd.read_csv(path, parse_dates=['date'])
    FEATURES[region] = frame
    print(f'{region}: {len(frame)} rows; {frame.date.min().date()} to {frame.date.max().date()}')
```

## Uttarakhand

```python
frame = FEATURES['uttarakhand']
display(frame.describe(include='all'))
display(frame.isna().sum().sort_values(ascending=False).to_frame('null_count'))
display(frame.fire_label.value_counts().rename_axis('fire_label').to_frame('days'))
frame[['temperature_2m_mean', 'precipitation_sum', 'relative_humidity_2m_mean', 'ndvi_mean', 'ndwi_mean']].hist(figsize=(12, 7))
plt.suptitle('Uttarakhand feature distributions')
plt.show()
```

## Southern California

```python
frame = FEATURES['california']
display(frame.describe(include='all'))
display(frame.isna().sum().sort_values(ascending=False).to_frame('null_count'))
display(frame.fire_label.value_counts().rename_axis('fire_label').to_frame('days'))
frame[['temperature_2m_mean', 'precipitation_sum', 'relative_humidity_2m_mean', 'ndvi_mean', 'ndwi_mean']].hist(figsize=(12, 7))
plt.suptitle('Southern California feature distributions')
plt.show()
```

## Southeastern Australia

```python
frame = FEATURES['australia']
display(frame.describe(include='all'))
display(frame.isna().sum().sort_values(ascending=False).to_frame('null_count'))
display(frame.fire_label.value_counts().rename_axis('fire_label').to_frame('days'))
frame[['temperature_2m_mean', 'precipitation_sum', 'relative_humidity_2m_mean', 'ndvi_mean', 'ndwi_mean']].hist(figsize=(12, 7))
plt.suptitle('Southeastern Australia feature distributions')
plt.show()
```

## Cross-Region Comparison

```python
comparison = pd.DataFrame({
    region: {
        'rows': len(frame),
        'start_date': frame.date.min().date(),
        'end_date': frame.date.max().date(),
        'positive_label_days': int(frame.fire_label.sum()),
        'positive_rate': frame.fire_label.mean(),
        'satellite_missing_days': int(frame.satellite_scene_count.isna().sum()),
        'weather_complete_days': int(frame[['temperature_2m_mean', 'precipitation_sum', 'relative_humidity_2m_mean']].notna().all(axis=1).sum()),
    }
    for region, frame in FEATURES.items()
}).T
display(comparison)

pd.concat(FEATURES, names=['region']).reset_index(level=0).boxplot(
    column=['temperature_2m_mean', 'precipitation_sum', 'relative_humidity_2m_mean'],
    by='region', figsize=(12, 5)
)
plt.suptitle('')
plt.show()
```
